```python
# Export .emd files (Thermo Scientific) to png format

%matplotlib qt

import hyperspy.api as hs
import numpy as np

import matplotlib.pyplot as plt

import glob
import os

# Place your data in folders relative to this script
folder = "./data/input"

# find all .emd files in the folder
files = glob.glob(os.path.join(folder, "*.emd"))

for f in files:
    print("Processing:", f)

    # Load the data
    d = hs.load(f)

    # Create a new figure for each plot
    plt.figure()
    d.plot()

    # Prepare output path
    png_path = f.replace(".emd", ".png")

    # Save as PNG
    plt.savefig(png_path, dpi=300)

    # Close the figure so memory doesn't explode
    plt.close()

print("Done!")
```
