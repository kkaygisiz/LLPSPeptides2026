```python
# Generate combinations with variable subsets of amino acids

import itertools
import csv

def generate_peptide_library(fixed_pattern, variable_positions, variable_letters, output_file):
    # Identify unique variable positions
    variable_positions_list = [pos for pos, char in enumerate(fixed_pattern) if char in variable_positions]
    
    # Generate all possible combinations of variable letters
    variable_combinations = itertools.product(*[variable_letters[fixed_pattern[pos]] for pos in variable_positions_list])
    
    # Generate peptide sequences by filling in the variable positions
    peptides = []
    for combo in variable_combinations:
        peptide = list(fixed_pattern)
        for i, pos in enumerate(variable_positions_list):
            peptide[pos] = combo[i]
        peptides.append("".join(peptide))
    
    # Write to CSV file
    with open(output_file, mode='w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(["Peptide"])
        for peptide in peptides:
            writer.writerow([peptide])
    
    print(f"Generated {len(peptides)} peptide sequences and saved to {output_file}")

# Define parameters
fixed_pattern = "XXK"  # Fixed design scheme
variable_positions = {'X': ['A', 'R', 'N', 'D', 'E', 'Q', 'G', 'H', 'L', 'K', 'M', 'F', 'P', 'S', 'T', 'W', 'Y', 'V']}
output_file = './data/Lib2_theoretical.csv'

# Generate the library
generate_peptide_library(fixed_pattern, variable_positions, variable_positions, output_file)
```
